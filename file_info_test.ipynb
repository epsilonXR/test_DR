```python
import torch
import numpy as np



import os
import warnings
warnings.filterwarnings("ignore")

# 尝试导入需要的库
try:
    import torch
except ImportError:
    torch = None

try:
    import numpy as np
except ImportError:
    np = None

def inspect_deep_npz(file_paths):
    print(f"🚀 开始深度结构扫描，共 {len(file_paths)} 个文件...\n")
    
    for i, file_path in enumerate(file_paths, 1):
        print(f"{'='*60}")
        print(f"📁 [{i}/{len(file_paths)}] 正在剖析: {file_path}")
        
        if not os.path.exists(file_path):
            print("❌ 文件不存在，跳过。")
            continue
            
        try:
            # 允许加载 Python 对象
            data = np.load(file_path, allow_pickle=True)
            keys = data.files
            print(f"✅ 确认是 NPZ 压缩包 | 共包含 {len(keys)} 个变量 (Keys):\n")
            
            for k in keys:
                item = data[k]
                print(f"  🔑 Key: '{k}'")
                print(f"     -> 外层类型: {type(item).__name__}")
                
                # 如果是标准的 NumPy 数组
                if isinstance(item, np.ndarray):
                    print(f"     -> 数组形状: {item.shape}")
                    print(f"     -> 数据类型: {item.dtype}")
                    
                    # ⚠️ 关键点：如果是包含对象的 0 维数组（通常用来存字典或列表）
                    if item.shape == () and item.dtype == 'O':
                        actual_content = item.item() # 提取里面的真实对象
                        print(f"     -> 🚨 这是一个特殊包装！里面实际藏着: {type(actual_content).__name__}")
                        
                        # 如果里面藏的是字典，把字典的键也打印出来
                        if isinstance(actual_content, dict):
                            dict_keys = list(actual_content.keys())
                            print(f"     -> 字典包含了 {len(dict_keys)} 个 Key，前几个是: {dict_keys[:5]}")
                        # 如果里面藏的是列表
                        elif isinstance(actual_content, list):
                            print(f"     -> 列表长度为 {len(actual_content)}，第一个元素类型: {type(actual_content[0]).__name__}")
                
                # 如果根本不是数组（比如直接存了个字符串）
                else:
                    print(f"     -> 值/摘要: {str(item)[:50]}...")
                    
                print("-" * 40)
                
            data.close()
            
        except Exception as e:
            print(f"❌ 解析该文件时出错: {e}")
        
        print("\n")

if __name__ == "__main__":
 
    file_paths = [r"\\cabinet\derivatives\DeepDR\FVs_biomed"
, r"\\cabinet\derivatives\DeepDR\FVs_ConceptCLIP"
, r"\\cabinet\derivatives\DeepDR\FVs_CONCH"
, r"\\cabinet\derivatives\DeepDR\FVs_CXRCLIP"
, r"\\cabinet\derivatives\DeepDR\FVs_DINOv2"
, r"\\cabinet\derivatives\DeepDR\FVs_EVA02-L-14-336"
, r"\\cabinet\derivatives\DeepDR\FVs_FLAIR"
, r"\\cabinet\derivatives\DeepDR\FVs_INViT-L-16"
, r"\\cabinet\derivatives\DeepDR\FVs_llava-med"
, r"\\cabinet\derivatives\DeepDR\FVs_llava-Mistral-7b"
, r"\\cabinet\derivatives\DeepDR\FVs_llava-vicuna-13b-hf"
, r"\\cabinet\derivatives\DeepDR\FVs_LLM2CLIP-EVA"
, r"\\cabinet\derivatives\DeepDR\FVs_LLM2CLIP-openai"
, r"\\cabinet\derivatives\DeepDR\FVs_medsiglip"
, r"\\cabinet\derivatives\DeepDR\FVs_MedTrinity"
, r"\\cabinet\derivatives\DeepDR\FVs_MONET"
, r"\\cabinet\derivatives\DeepDR\FVs_resnet50"
, r"\\cabinet\derivatives\DeepDR\FVs_SigLIP"
, r"\\cabinet\derivatives\DeepDR\FVs_swin"
, r"\\cabinet\derivatives\DeepDR\FVs_UNI"
, r"\\cabinet\derivatives\DeepDR\FVs_VGG16"
, r"\\cabinet\derivatives\DeepDR\FVs_ViT-B-16"
, r"\\cabinet\derivatives\DeepDR\FVs_ViT-g-14"
, r"\\cabinet\derivatives\DeepDR\FVs_ViT-L-14"]
    
    inspect_deep_npz(file_paths)
```

```python
import os
import torch
import warnings

# 忽略一些不必要的警告
warnings.filterwarnings("ignore")

def inspect_torch_file(file_path):
    """深度解析单个 PyTorch 文件的数据结构"""
    try:
        # weights_only=False 允许加载包含字典或列表的复杂结构
        data = torch.load(file_path, map_location='cpu', weights_only=False)
        
        print(f"   ✅ 读取成功！顶层数据类型: {type(data).__name__}")
        
        # 情况 1: 如果读出来是一个字典 (最常见的情况)
        if isinstance(data, dict):
            keys = list(data.keys())
            print(f"   -> 这是一个字典，共包含 {len(keys)} 个键 (Keys)。")
            print(f"   -> 前几个键名: {keys[:5]}")
            
            # 抽查第一个键的内容，看它是不是特征矩阵
            first_key = keys[0]
            first_val = data[first_key]
            print(f"   -> 🔍 抽查 ['{first_key}']:")
            print(f"      - 数据类型: {type(first_val).__name__}")
            
            if hasattr(first_val, 'shape'):
                print(f"      - 矩阵形状 (Shape): {first_val.shape}")
                if hasattr(first_val, 'dtype'):
                    print(f"      - 数据精度 (Dtype): {first_val.dtype}")
            elif isinstance(first_val, dict):
                 print(f"      - 这是一个嵌套字典，内部包含 {len(first_val)} 个键。")
            elif isinstance(first_val, list):
                 print(f"      - 这是一个列表，长度为 {len(first_val)}。")
                 
        # 情况 2: 如果读出来直接就是一个纯张量/矩阵
        elif hasattr(data, 'shape'):
             print(f"   -> 这是一个纯张量 (Tensor)！")
             print(f"   -> 矩阵形状 (Shape): {data.shape}")
             if hasattr(data, 'dtype'):
                 print(f"   -> 数据精度 (Dtype): {data.dtype}")
                 
        # 情况 3: 如果是列表
        elif isinstance(data, list):
             print(f"   -> 这是一个列表 (List)，包含 {len(data)} 个元素。")
             if len(data) > 0 and hasattr(data[0], 'shape'):
                 print(f"   -> 第一个元素的形状 (Shape): {data[0].shape}")
                 
        else:
             print(f"   -> 其他数据结构，值摘要: {str(data)[:50]}...")
             
    except Exception as e:
        print(f"   ❌ PyTorch 读取失败，错误信息: {e}")


def batch_process_directory(target_dir):
    """自动扫描目录并批量处理"""
    if not os.path.exists(target_dir):
        print(f"❌ 找不到目录: {target_dir}")
        print("请检查你的网络驱动器 (cabinet) 是否已连接。")
        return

    # 自动获取目录下所有以 FVs_ 开头的文件
    print("正在扫描目录...")
    all_files = os.listdir(target_dir)
    target_files = [f for f in all_files if f.startswith("FVs_")]
    
    # 按字母顺序排个序，方便查看
    target_files.sort()

    print(f"🚀 扫描完毕！共找到 {len(target_files)} 个特征文件。开始逐一解析...\n")

    for i, file_name in enumerate(target_files, 1):
        file_path = os.path.join(target_dir, file_name)
        print(f"{'='*60}")
        print(f"📁 [{i}/{len(target_files)}] 正在分析: {file_name}")
        
        inspect_torch_file(file_path)
        print("\n")


if __name__ == "__main__":
    # 指定你的网络共享文件夹路径 (记得保留 r 前缀)
    DIRECTORY_PATH = r"\\cabinet\derivatives\DeepDR"
    
    batch_process_directory(DIRECTORY_PATH)
```

```python
import torch
import os

def analyze_keys(file_path):
    print(f"正在读取文件: {file_path} ...\n")
    
    try:
        data = torch.load(file_path, map_location='cpu', weights_only=False)
        keys = list(data.keys())
        
        # 1. 基础展示
        print("=" * 50)
        print(f"🔑 总键数 (Total Keys): {len(keys)}")
        print(f"   -> 前 15 个键: {keys[:15]}")
        print(f"   -> 最后 5 个键: {keys[-5:]}")
        print("=" * 50)
        
        # 2. 深度规律分析
        patient_ids = set()
        eye_positions = {}
        
        for k in keys:
            # 假设格式是 "1_l1" -> 以 "_" 分割
            parts = k.split('_')
            if len(parts) >= 2:
                p_id = parts[0]
                pos = parts[1]
                
                patient_ids.add(p_id)
                # 统计各个位置（如 l1, r1）的数量
                eye_positions[pos] = eye_positions.get(pos, 0) + 1
        
        print("\n📊 数据集内在结构分析：")
        print(f"👨‍⚕️ 独立患者总数: {len(patient_ids)} 人")
        print("👁️ 各眼睛位置/视角的图片数量分布:")
        for pos, count in sorted(eye_positions.items()):
            # 简单美化一下输出：l 代表左眼(Left)，r 代表右眼(Right)
            side = "左眼(Left)" if pos.startswith('l') else "右眼(Right)" if pos.startswith('r') else "未知"
            print(f"   - [{pos}] ({side}): {count} 张")
            
        # 3. 导出到文件（如果你想看完整的 1200 个）
        output_txt = "FVs_CONCH_keys.txt"
        with open(output_txt, "w", encoding="utf-8") as f:
            for k in keys:
                f.write(k + "\n")
                
        print("\n" + "=" * 50)
        print(f"✅ 完整的 1200 个键名已导出到当前目录的: {output_txt}")
        print("你可以直接用记事本打开它查看所有详细编号。")

    except Exception as e:
         print(f"❌ 读取失败: {e}")

if __name__ == "__main__":
    # 直接指定你的 FVs_CONCH 文件路径
    TARGET_FILE = r"\\cabinet\derivatives\DeepDR\FVs_CONCH"
    analyze_keys(TARGET_FILE)
```

```python
import torch
import numpy as np
import os

def build_multimodel_dataset(target_dir):
    all_files = sorted([f for f in os.listdir(target_dir) if f.startswith("FVs_")])
    
    # 1. 制作“标准点名册” (以第一个文件为准)
    first_file = os.path.join(target_dir, all_files[0])
    first_data = torch.load(first_file, map_location='cpu', weights_only=False)
    
    # 直接拿到原本就已经完美的顺序！
    master_keys = list(first_data.keys()) 
    print(f"📋 已生成标准点名册，共 {len(master_keys)} 个样本。首个样本: {master_keys[0]}, 末尾样本: {master_keys[-1]}")
    
    # 2. 准备一个大字典，用来存所有模型的矩阵
    all_matrices = {}
    
    # 3. 挨个点名提取
    for file_name in all_files:
        file_path = os.path.join(target_dir, file_name)
        data = torch.load(file_path, map_location='cpu', weights_only=False)
        
        feature_list = []
        # 严格按照 master_keys 的顺序去取特征！
        for k in master_keys:
            vec = data[k]
            if isinstance(vec, torch.Tensor):
                vec = vec.numpy()
            feature_list.append(vec.flatten())
            
        # 堆叠成矩阵并保存
        X = np.vstack(feature_list)
        # 去掉 'FVs_' 前缀，用纯模型名做 key
        model_name = file_name.replace("FVs_", "") 
        all_matrices[model_name] = X
        
        print(f"✅ [{model_name}] 矩阵构建完成 -> 形状: {X.shape}")
        
    return all_matrices, np.array(master_keys)

if __name__ == "__main__":
    DIRECTORY_PATH = r"\\cabinet\derivatives\DeepDR"
    
    # 运行一次，拿回所有模型的矩阵和共用的标签！
    matrices, labels = build_multimodel_dataset(DIRECTORY_PATH)
    
    print("\n🎉 大功告成！你可以这样调用它们：")
    print(f"-> ResNet50 的特征矩阵: matrices['resnet50'].shape")
    print(f"-> ViT-L-14 的特征矩阵: matrices['ViT-L-14'].shape")
    print(f"-> 它们共同对应的标签: labels.shape")
```
